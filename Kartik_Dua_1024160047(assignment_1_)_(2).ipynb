```python
import random
from collections import Counter

roll = 1024160131
random.seed(roll)

nums = [random.randint(100, 900) for _ in range(100)]

print("List:")
print(nums)

odd = [i for i in nums if i % 2 != 0]
print("\nOdd Count:", len(odd))

even = [i for i in nums if i % 2 == 0]
print("Even Count:", len(even))

def prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

primes = [i for i in nums if prime(i)]
print("Prime Count:", len(primes))
print("Prime Numbers:", primes)

c = Counter(nums)
most = c.most_common(1)[0]
print("Most Frequent Number:", most[0])
print("Occurrences:", most[1])
```

```python
digits = [1,0,2,4,1,6,0,1,3,1]

A = {i*7 for i in digits}
B = {i*9 for i in digits}

print("A =", A)
print("B =", B)

print("Union =", A.union(B))

print("Intersection =", A.intersection(B))

print("A-B =", A.difference(B))
print("B-A =", B.difference(A))

print("Symmetric Difference =", A.symmetric_difference(B))

print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

x = int(input("Enter value to remove from A: "))
A.discard(x)
print("Updated A:", A)
```

```python
my_dict = {
    "name": "Kartik Dua",
    "roll_no": "1024160047",
    "branch": "CSE",
    "age": 20,
    "city": "Panipat"
}

my_dict["location"] = my_dict.pop("city")
print(my_dict)

my_dict["cgpa"] = 8.0
print(my_dict)

my_dict["age"] += 1
print(my_dict)

d1 = my_dict.copy()
d1.pop("branch")
print(d1)

d2 = my_dict.copy()
del d2["branch"]
print(d2)

for k, v in my_dict.items():
    print(k, "->", v)

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key not found")

friend_dict = {
    "name": "xyz",
    "roll_no": "1234567890",
    "branch": "ECE",
    "age": 21,
    "city": "Delhi"
}

merged = {**my_dict, **friend_dict}
print(merged)

string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_dict)
```

```python
import pandas as pd

data = {
    "Name": ["kartik", "Rahul", "Priya", "Karan"],
    "Age": [20, 21, 20, 22],
    "Marks": [91, 78, 95, 67]
}

df = pd.DataFrame(data)

print(df)
print("\nAverage marks:", df["Marks"].mean())
print("\nTopper:")
print(df.loc[df["Marks"].idxmax()])
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Kartik", "Rahul", "Priya", "Karan"],
    "Marks": [91, np.nan, 95, 72],
    "Age": [20, 21, np.nan, 22]
})

print(df)

print("\nMissing values:")
print(df.isnull().sum())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Age"] = df["Age"].fillna(df["Age"].median())

print("\nAfter filling:")
print(df)
```
