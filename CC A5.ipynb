```python
 x = np.array([1,2,3,4,5,1,2,1,1,1])

 values,counts= np.unique(x,return_counts=True)
 mostfrequency=values[np.argmax(counts)]
 indices=np.where(x==mostfrequency)

 print(mostfrequency)
 print(np.max(counts))
 print(indices)

 y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

most_frequent = values[np.argmax(counts)]

indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Frequency:", np.max(counts))
print("Indices:", indices)

```

```python
import numpy as np

ucs420_Arpit = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Arpit)

# Mean
print("Mean:", np.mean(ucs420_Arpit))

# Median
print("Median:", np.median(ucs420_Arpit))

# Maximum
print("Maximum:", np.max(ucs420_Arpit))

# Minimum
print("Minimum:", np.min(ucs420_Arpit))

# Unique elements
print("Unique elements:", np.unique(ucs420_Arpit))

# Reshape into 4 rows and 3 columns
reshaped_ucs420_Arpit = ucs420_Arpit.reshape(4, 3)

print("Reshaped Array:")
print(reshaped_ucs420_Arpit)

# Resize into 2 rows and 3 columns
resized_ucs420_Arpit = np.resize(ucs420_Arpit, (2, 3))

print("Resized Array:")
print(resized_ucs420_Arpit)
```
